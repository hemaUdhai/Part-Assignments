```python
## Task 1: While Loop & Control Statements -Number Guessing Game
```

```python
import random

secret_number = random.randint(1, 11)  # it Generates random number between 1 to 10

attempts = 3

while attempts > 0:
    num = int(input("Guess the number (between 1 and 10): "))

    # it Checks if the guess is out of range or not
    if num < 1 or num > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    # it Checks if the guess is correct or not
    if num == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    elif num  > secret_number:
        print("Your guess is greater than the secret number.")
    else:
        print("Your guess is lower than the secret number.")

    attempts -= 1
    print("Attempts remaining:", attempts)

else:
    print("Better luck next time!")
    print("The secret number was:", secret_number)
```

```python
## Task 2:- For Loop - Multiplication Table Generator
```

```python

number = int(input("Enter the number for which you want the multiplication table: "))

# Generate multiplication table from 1 to 10
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")
```

```python
## Task 3: Function- BMI Calculator

def calculate_BMI(weight, height):
    BMI = weight / (height ** 2)
    return BMI

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

BMI = calculate_BMI(weight, height)

print(f"Your BMI is:{round(BMI, 2)}")
```
