```python
## List (Creation, Modifi cation and Access):
```

```python
##1. List Creation:
## a. Create age_list with five integer elements

age_list = [25,28,32,40,55,60]

print(age_list)
```

```python
## b. Create name_list with five string elements

name_list = ["Hema", "Arun", "Deepika", "Divya", "Jai"]

print(name_list)
```

```python
# 2. List Operations / Modifications:

# a. Append the string "Yazhini" to name_list.

name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert the element 30 at index 2 in age_list.

age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove the string "Yazhini" from name_list.

name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element from age_list.

age_list.pop()
print("After pop:", age_list)

# e. Extend the age_list with additional ages [29, 30, 26].

age_list.extend([29, 30, 26])
print("After extend :", age_list)

# f. Sort age_list in descending order.

age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Find Max age, Min age and sum of all ages from age_list.

print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# 3. Accessing List Elements:

# a. Print the first element

print("First element:", name_list[0])

# b. Print the last element

print("Last element:", name_list[-1])

# c. Print elements from index 2 to index 4

print("Elements from index 2 to 4:", name_list[2:5])

# d. Print name_list in reverse order

print("Reverse order:", name_list[::-1])
```

```python
# Dictionary (Creation, Modifi cation and Access):

# a. Create a dictionary with five students and their marks

student_marks = {
    "Hema": 95,
    "Arun": 78,
    "Deepika": 92,
    "Divya": 75,
    "Jai": 55
}
```

```python
print("Student Marks:", student_marks)
```

```python
# b. Access and print the mark of a specific student

print("Deepika's mark:", student_marks["Deepika"])
```

```python
# c. Add a new student "Janani" with a mark of 80

student_marks["Janani"] = 80
print("Student Marks:", student_marks)
```

```python
# d. Update the mark of an older student to 82

student_marks["Arun"] = 82
print("Student Marks :", student_marks)
```

```python
# e. Print all keys, values, and key-value pairs

print("Keys:", student_marks.keys()) # Gets all student names
print("Values:", student_marks.values()) # Gets all marks
print("Key-Value pairs:", student_marks.items()) # Gets both names and marks
```

```python
### Sets (Operations):

# a. Create a set and analyze the output

my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

```python
# Explanation:

# A set does not allow duplicate values.
# The values 'a' and 'i' appear multiple times in the input, but they are stored only once.
# So, the set contains only 5 unique vowels.
# The order of elements in a sets are unordered in Python.
```

```python
# b. Attempt to change my_set[4]

my_set[4] = 's'
```

```python
# Explanation:

# Sets are unordered collections, so they do not have indexes like lists.
# is not valid because we cannot access a set using an index.
```

```python
## c. Create two sets:


set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
## d. Find Union and Intersection

# Union
union_set = set1.union(set2)

# Intersection
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```

```python
### Operators & Conditional Statements :
# (IF, ELIF, ELSE)
# Performance Category Program:

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent performance! Keep up the good work.")

elif score >= 4 and score <= 7:
    print("Performance Category: Average")
    print("Good effort! Keep practicing to improve your performance.")

else:
    print("Performance Category: Below Average")
    print("Need to improve your performance. Consistent practice will lead to better results.")
```
