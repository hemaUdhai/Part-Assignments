```python
# String Slicing and Indexing

name = input("Enter your Name: ")

# Create the concatenated string
string = "Hello " + name + ", Welcome to Python programming"

print("\nExisting String:", string)

# a. Print the first character
print("a.First character:", string[0])

# b. Print the last character
print("b.Last character:", string[-1])

# c. Print the first 5 characters
print("c.First 5 characters:", string[:5])

# d. Print the last 11 characters
print("d.Last 11 characters:", string[-11:])

# e. Print the string in reverse
print("e.Reverse:", string[::-1])

# f. Print the word "Python"
start = string.index("Python")
print("f.Word Python:", string[start:start + 6])
```

```python
# String Methods

strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase
print("a. Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("b. Lowercase:", strM.lower())

# c. Capitalize the sentence
print("c. Capitalize:", strM.capitalize())

# d. Count the occurrences of character 't'
print("d. Count of 't':", strM.count('t'))

# e. Replace "Python" with "Data Analytics"
print("e. Replaced string:", strM.replace("Python", "Data Analytics"))
```

```python
# Tuples - Creation, Modification and Access

# Create the first tuple
t1 = (10, 20, 30)

# Create the second tuple
t2 = (40, 45, 50, 60)

# a. Concatenate the two tuples
t_combine = t1 + t2
print("a. Combined tuple:", t_combine)

# b. Repeat the elements of t_combine 3 times
t_repeat = t_combine * 3
print("b. Repeated tuple:", t_repeat)

# c. Access the 3rd element from t_combine
print("c. 3rd element:", t_combine[2]) # the index starts with 0th value

# d. Access the first three elements from t_combine
print("d. First three elements:", t_combine[:3])

# e. Access the last three elements from t_combine
print("e. Last three elements:", t_combine[-3:])
```
