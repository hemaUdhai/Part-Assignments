```python
##  Numpy Array Operations:

## 1. Create a 1D NumPy Array

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)
```

```python
## 2. Inspect the Array Properties


temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Shape:", temperatures_w1.shape) # It returns the dimensions of the array

print("Data Type:", temperatures_w1.dtype)  # It returns the type of elements stored in the array

print("Number of Elements:", temperatures_w1.size)  # it returns the total number of elements

print("Array Shape:", np.shape(temperatures_w1))   ## it delivers the shape of the array
```

```python
## 3. Array Operations

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

temperatures_Fahrenheit = (temperatures_w1 * 9/5) + 32   # # Convert Celsius to Fahrenheit
print("Temperatures in Fahrenheit:", temperatures_Fahrenheit)


print("Maximum Temperature:", np.max(temperatures_w1)) # it finds the maximum temperature that is highest value.


print("Minimum Temperature:", np.min(temperatures_w1))  # it finds the minimum temperature, the lowest value.


print("Mean Temperature:", np.mean(temperatures_w1)) # it finds the mean temperature,calculates the average value.
```

```python
## 4. Array Slicing and Indexing
## syntax for slicing is:  array[start:stop]
## start: The index where extraction begins & stop: The index where extraction ends, but that index is not included.

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("First Three Days:", temperatures_w1[:3]) # it extracts temperatures for the first three days

print("Weekend Temperatures:", temperatures_w1[-2:]) # it extracts temperatures for the weekend (last two days)

print("Middle Three Days:", temperatures_w1[2:5])  # it extracts temperatures for the middle three days
```

```python
## 5. Create a 2D NumPy Array
# A 2D NumPy array contains rows and columns.each row represents the temperatures recorded during one week.

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2 Each row has 7 elements, and the array has 2 rows.
])

print("Temperatures for Two Weeks:")
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array
# Inspect the Array Properties

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Shape:", temperatures.shape)  # the dimensions of the array, 2 rows and 7 columns

print("Data Type:", temperatures.dtype) ## it Stores decimal numbers

print("Total Number of Elements:", temperatures.size) # the total number of elements in 2D array

# Extract Temperatures for Each Week

print("Week 1 Temperatures:", temperatures[0]) # Week 1

print("Week 2 Temperatures:", temperatures[1])  # week 2

# Extract Weekend Temperatures for Both Weeks, The weekend temperatures are the last two values in each row.

print("Weekend Temperatures for Both Weeks:")
print(temperatures[:, -2:]) # : it selects all rows from both weeks, -2: it selects the last two columns in the both weekend.
```

```python
# Pandas Series
## 1. Creating a Pandas Series
# Pandas Series is a one-dimensional data structure that stores values along with their index labels

import pandas as pd

marks = pd.Series(              # it creates a Pandas Series, represents the marks.
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']  # index assigns custom labels to the marks.
)

print(marks)
```

```python
# 2. Indexing and Slicing

# to Access the 1st Rank Student Using Integer Index Position

print("1st Rank Mark:", marks.iloc[0]) # iloc[0] retrieves the first value because Python indexing starts at 0

# Retrieve the Top 3 Ranks Using loc

print(marks.loc[['Rank1', 'Rank2', 'Rank3']]) # loc accesses values using their index labels to retrieve the marks of the top three students.

# Retrieve the 3rd-Ranked Student Using iloc

print("3rd Rank Mark:", marks.iloc[2]) # iloc[2] retrieves the third value because the index positions begin at 0

# Filter Ranks Where Marks Are Greater Than 90.

print(marks[marks > 90])  # checks which marks are greater than 90 and returns only the values that satisfy the condition
```

```python
# 3. Manipulating Pandas Series
# to modify the 1st Rank Student's Mark to 100

marks.loc['Rank1'] = 100 # it updates the mark with the Rank 1 to 100.

print(marks)
```

```python
# Remove the Last Rank from the Series

marks = marks.drop('Rank5')  # it removes the entry with Rank5

print(marks)

# Compute CGPA by Dividing Each Mark by 10

cgpa = marks / 10  # it Divides a Pandas Series by 10 applies the operation to every element.e.g, 100 / 10 = 10.0

print("CGPA:")
print(cgpa)
```

```python
# Pandas DataFrame
# Creating a Pandas DataFrame -- Pandas DataFrame is a two-dimensional data structure arranged in rows and columns, similar to an Excel table.

import pandas as pd

transactions = pd.DataFrame({                                           # it creates the Datafframe
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
# 2. Data Exploration
## to Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

print(transactions)
```

```python
print(transactions.head())   # the head()function displays the first five rows by default.
```

```python
print(transactions.tail())   # the tail()function displays the last five rows by default.
```

```python
print("Shape:", transactions.shape)
```

```python
print("Column Names:", transactions.columns)  # the (columns) returns all column names in the DataFrame.
```

```python
transactions.info()   # info() displays the DataFrame's structure.
```

```python
#  Display Only ProductCategory and Amount

print(transactions[['ProductCategory', 'Amount']])  # we should use double square brackets when selecting multiple columns.it display only the selected columns
```

```python
# Retrieve the Last Three Columns Using iloc

print(transactions.iloc[:, -3:])    # : selects all rows, -3: selects the last three columns.
```

```python
#  Filter Rows Where Region Is North and Amount Is Greater Than 200

filtered_transactions = transactions[
    (transactions['Region'] == 'North') &    ## it selects rows from the North region , & combines both conditions, so both must be true.we should use parentheses around each condition when combining conditions with &.
    (transactions['Amount'] > 200)           ## it selects amounts greater than 200.
]

print(filtered_transactions)
```

```python
## to Find Value Counts for ProductCategory

print(transactions['ProductCategory'].value_counts())   # it counts how many times each category appears.
```

```python
## Find Unique Values in the Region Column

print(transactions['Region'].unique())   # The unique()function returns the distinct region values without duplicates.
```

```python
## Group by Region and Calculate Mean Amount

print(transactions.groupby('Region')['Amount'].mean())  # it groups transactions by region and amount columns,mean() calculates the average amount for each region.
```

```python
## 3. Manipulating the DataFrame
## To Modify Amount for TransactionID 102 to 165

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165  # it identifies the required transaction.

print(transactions)
```

```python
## Add a new column 'Discount' by calculating 10% of the 'Amount'.

transactions['Discount'] = transactions['Amount'] * 0.10   # it declares the new column named Discount and calculates 10% of each transaction amount.

print(transactions)
```

```python
## Remove the Row with TransactionID 109

transactions = transactions.drop(                              # .drop() removes that row.
    transactions[transactions['TransactionID'] == 109].index   # transactions['TransactionID'] == 109 identifies the row, retrieves its index label.
)

print(transactions)                                             # TransactionID 109 is now removed from the DataFrame.
```

```python
#  Delete the Discount Column

transactions = transactions.drop(columns=['Discount']) # drop(columns=['Discount']) it removes the Discount column

print(transactions)
```
